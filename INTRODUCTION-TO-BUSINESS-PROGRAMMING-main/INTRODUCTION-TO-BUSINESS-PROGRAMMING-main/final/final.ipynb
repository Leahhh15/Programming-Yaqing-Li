### 문제. cpi 변동을 기준으로 거래총가치(sales_total)와 거래가중앙값(price_med)을 아래 수식을 참고하여 파생변수를 만드세요. txhousing 데이터의 앞부분을 확인해보세요.


cpi_latest = 238.654

sales_total_adj = sales_total * cpi_latest / cpi (소수점 한자리까지 표시되도록 하세요.)

price_med_adj = price_med * cpi_latest / cpi (소수점 한자리까지 표시되도록 하세요.)


```python
cpi_latest = 238.654
# 거래총가치 조정 파생변수 생성
txhousing['sales_total_adj'] = (txhousing['sales_total'] * cpi_latest / txhousing['cpi']).round(1)
# 거래가중앙값 조정 파생변수 생성
txhousing['price_med_adj'] = (txhousing['price_med'] * cpi_latest / txhousing['cpi']).round(1)
# 결과 확인
txhousing.head()
```

### 문제. med_tx 데이터의 act_cnt_med의 극단치를 제거하여 med_tx_new에 할당하고, 다시 확인해보세요.

```python
# med_tx.act_cnt_med 변수의 1사분위수 계산
q1 = np.percentile(med_tx.act_cnt_med, 25)
# med_tx.act_cnt_med 변수의 3사분위수 계산
q3 = np.percentile(med_tx.act_cnt_med, 75)
# IQR 계산
iqr = q3 - q1
# 이상치 상한선 계산 (상자 수염 그림 기준)
upper_bound = q3 + (iqr * 1.5) 
# 이상치 하한선 계산 (상자 수염 그림 기준)
lower_bound = q1 - (iqr * 1.5)
# 이상치를 제외한 데이터만 선택하여 새로운 med_tx 데이터프레임 생성
med_tx = med_tx[med_tx.act_cnt_med <= upper_bound]
# 이상치 제거 후 med_tx.act_cnt_med 변수의 상자 수염 그림
sns.boxplot(x = med_tx.act_cnt_med)
```

### 문제. 아래 코드를 실행 후 생성된 파생변수를 파악하여 간단히 주석으로 남기세요.

```python
txhousing['act_cnt_ind'] = np.where(txhousing['act_cnt'] <= txhousing['act_cnt_med'], 0, 1)
txhousing['sales_cnt_ind'] = np.where(txhousing['sales_cnt'] <= txhousing['sales_cnt_med'], 0, 1)
condition_list = [((txhousing['act_cnt_ind'] == 1) & (txhousing['sales_cnt_ind'] == 0)),
                 ((txhousing['act_cnt_ind'] == 0) & (txhousing['sales_cnt_ind'] == 0)),
                 ((txhousing['act_cnt_ind'] == 1) & (txhousing['sales_cnt_ind'] == 1)),
                 ((txhousing['act_cnt_ind'] == 0) & (txhousing['sales_cnt_ind'] == 1))]
choice_list = ['1', '2', '3', '4']
txhousing['market_hotness'] = np.select(condition_list, choice_list, default='Not Specified')
txhousing['market_hotness'].value_counts()
```

(여기에 파악한 내용을 남기세요.)
act_cnt_ind 변수는 act_cnt 값이 act_cnt_med 값보다 작거나 같은 경우 0으로 설정되고, 그렇지 않은 경우 1로 설정된다.
sales_cnt_ind 변수는 sales_cnt 값이 sales_cnt_med 값보다 작거나 같은 경우 0으로 설정되고, 그렇지 않은 경우 1로 설정된다.
market_hotness 변수는 위에서 생성된 act_cnt_ind와 sales_cnt_ind 변수의 조합에 따라 다른 값을 가지도록 설정된다. 각 조건에 따라 1, 2, 3, 4 값 중 하나가 선택되며, 조건에 해당하지 않는 경우 'Not Specified'로 설정된다.
market_hotness 변수의 값에 대한 빈도를 계산하여 출력합니다. 결과는 각 값(1, 2, 3, 4)의 개수를 나타낸다.
따라서, 위의 코드는 act_cnt_ind, sales_cnt_ind, market_hotness 세 개의 파생변수를 생성하고, market_hotness 값에 따라 데이터를 분류하는 작업을 수행한다.

### 문제. market_hotness 변수의 값들을 순서대로 아래와 같이 값을 변경하고 다시 고유값별 빈도를 확인하세요.
|||
|--|--|
|1 | Very_Low|
|2 | Low|
|3 | Average|
|4 | High|

```python
# # market_hotness 변수 값들을 순서대로 변경
txhousing['market_hotness'] = txhousing['market_hotness'].map({'1': 'Very_Low',
                                                               '2': 'Low',
                                                               '3': 'Average',
                                                               '4': 'High'})
# 고유값별 빈도 확인
txhousing['market_hotness'].value_counts()
```
