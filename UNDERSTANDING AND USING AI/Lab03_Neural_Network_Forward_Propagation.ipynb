```python
# 3.20

def init_network():
    network = {}
    network['W1'] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    network['W2'] = np.array([[0.1, 0.4], [0.2, 0.4], [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    network['W3'] = np.array([[0.1, 0.3], [0.2, 0.4]])
    network['b3'] = np.array([0.1, 0.2])

    return network

def forward( network, x ):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = identity_function(a3)

    return y

network = init_network()
x = np.array([1.0, 0.5])
y = forward( network, x )
print(y)
```

```python
# 3.26


# 1D 입력을 처리할 수 있는 softmax

def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a

    return y
```

2D 배열을 지원하는 softmax 함수에 대한 상세한 설명은 아래 링크에서 AI03_Softmax.ipynb 참조

https://colab.research.google.com/drive/15M5jJZ_jk9FiAHHo4M5WULA0y--vOaua?usp=sharing


```python
# 2D 입력도 처리할 수 있는 softmax

def softmax(a):
    if a.ndim >= 2:
        c = np.max( a, axis=1, keepdims=True )
        exp_a = np.exp(a - c)
        sum_exp_a = np.sum( exp_a, axis=1, keepdims=True )
    else:
        c = np.max(a)
        exp_a = np.exp(a - c)
        sum_exp_a = np.sum(exp_a)

    return exp_a / sum_exp_a
```

```python
# 3.32

import pickle

def get_data():

    return x_test / 255, t_test

def init_network():
    with open('sample_weights.pkl', 'rb') as f:
        network = pickle.load(f)

    return network

def predict(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']
    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = softmax(a3)

    return y
```

```python
# 3.33

x, t = get_data()
network = init_network()
accuracy_count = 0
for i in range(len(x)):
    y = predict(network, x[i])
    p = np.argmax(y)
    if p == t[i]:
        accuracy_count += 1
print( 'Accuracy: {}'.format(accuracy_count / len(x) ) )
```

```python
# 3.35

x, t = get_data()
network = init_network()

batch_size = 100
accuracy_count = 0

for i in range(0, len(x), batch_size):
    x_batch = x[i:i+batch_size]
    y_batch = predict(network, x_batch)
    p = np.argmax(y_batch, axis=1)

# print(p)
# print( t[i:i+batch_size] )

    accuracy_count += np.sum( p == t[i:i+batch_size] )

print( 'Accuracy: {}'.format(accuracy_count / len(x) ) )
```
